```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import io, color, filters, feature
import skimage.filters.rank
from skimage.morphology import disk

# ===================== 1. LOAD GAMBAR =====================
# Ganti 'gambar.jpg' dengan nama file dari HP kamu
image = io.imread('kucing.jpg')

# ===================== 2. KONVERSI KE GRAYSCALE =====================
gray = color.rgb2gray(image)

# ===================== 3. DETEKSI TEPI =====================

# Sobel
sobel = filters.sobel(gray)

# Roberts
roberts = filters.roberts(gray)

# Prewitt
prewitt = filters.prewitt(gray)

# Kirsch (pakai pendekatan morfologi → perlu uint8)
gray_uint8 = (gray * 255).astype('uint8')
kirsch = skimage.filters.rank.gradient(gray_uint8, disk(1))

# Canny
canny = feature.canny(gray, sigma=1)

# ===================== 4. VISUALISASI =====================
fig, ax = plt.subplots(2, 3, figsize=(12, 8))

# Baris 1
ax[0, 0].imshow(gray, cmap='gray')
ax[0, 0].set_title("Grayscale")
ax[0, 0].axis("off")

ax[0, 1].imshow(sobel, cmap='gray')
ax[0, 1].set_title("Sobel")
ax[0, 1].axis("off")

ax[0, 2].imshow(roberts, cmap='gray')
ax[0, 2].set_title("Roberts")
ax[0, 2].axis("off") 

# Baris 2
ax[1, 0].imshow(prewitt, cmap='gray')
ax[1, 0].set_title("Prewitt")
ax[1, 0].axis("off")

ax[1, 1].imshow(kirsch, cmap='gray')
ax[1, 1].set_title("Kirsch")
ax[1, 1].axis("off")

ax[1, 2].imshow(canny, cmap='gray')
ax[1, 2].set_title("Canny")
ax[1, 2].axis("off")

plt.tight_layout()
plt.show()
```
